```python
th = {
 "00:0f:00:70:91:0a": {"co":0.009254,"smoke":0.033069,"lpg":0.012019,"temp":20.1},
 "1c:bf:ce:15:ec:4d": {"co":0.005006,"smoke":0.020572,"lpg":0.007707,"temp":30.3},
 "b8:27:eb:bf:9d:51": {"co":0.007189,"smoke":0.027209,"lpg":0.010014,"temp":23.6},
}
for dev, lims in th.items():
    sub = df[df.device == dev]
    hits = {m: int((sub[m] > t).sum()) for m, t in lims.items()}
    print(dev, len(sub), hits)
```
